```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../homeworks/hw_7_Darya/passengers.csv')
```

```python
X = df.drop(["Unnamed: 0", "Survived"], axis=1)
# X = df[['Pclass_3']]
y = df['Survived']
clf = LogisticRegression(random_state=0, max_iter=1000).fit(X, y)
clf.predict(X.iloc[:100, :])
clf.predict_proba(X.iloc[:100, :])
```

```python
print(f"Precision: {precision_score(y, clf.predict(X))}")
print(f"Recall: {recall_score(y, clf.predict(X))}")
print(f"F1_score: {f1_score(y, clf.predict(X))}")
print(f"Accuracy: {accuracy_score(y, clf.predict(X))}")
```

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter
```

```python
file_path = "creditcard.csv"

df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "mlg-ulb/creditcardfraud",
  file_path,
)

print("First 5 records:", df.head())
print("Description:", df.describe())
```

```python
X = df.drop("Class", axis=1)
y = df['Class']
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
clf = LogisticRegression(random_state=0, max_iter=1000).fit(x_train, y_train)
clf.score(x_train, y_train)
```

```python
y_pred = clf.predict(x_train)
```

```python
print(f"Predict: {precision_score(y_train, y_pred)}")
print(f"Recall: {recall_score(y_train, y_pred)}")
print(f"F1_score: {f1_score(y_train, y_pred)}")
```

```python
import numpy as np
np.set_printoptions(suppress=True, precision=10)
clf.predict_proba(x_train.iloc[:100, :])
```

```python
y_test_pred = clf.predict(x_test)
test_score = clf.score(x_test, y_test)
y_test_proba = clf.predict_proba(x_test)

print(f"Model score on test set (accuracy): {test_score}")
print("Predictions on first 100 test samples:", y_test_pred[:100])
print("Probabilities on first 100 test samples:", y_test_proba[:100])
```

```python
print(f"Predict: {precision_score(y_test, y_test_pred)}")
print(f"Recall: {recall_score(y_test, y_test_pred)}")
print(f"F1_score: {f1_score(y_test, y_test_pred)}")
```

```python
y_pred_proba = clf.predict_proba(x_train)

y_pred_thresholded = (y_pred_proba[:, 1] >= 0.99971).astype(int)
```

```python
print(f"Precision with new threshold: {precision_score(y_train, y_pred_thresholded)}")
print(f"Recall with new threshold: {recall_score(y_train, y_pred_thresholded)}")
print(f"F1_score with new threshold: {f1_score(y_train, y_pred_thresholded)}")
```
